# **Agente LLM: Ciclo iterativo con objetivo**

Un agente LLM es un sistema que percibe información, toma decisiones, ejecuta acciones e itera hasta cumplir un objetivo (Diap 8, módulo 2)

````python
# Simulación conceptual de un agente (Por ahora sin herramientas)
# Este ejemplo muestra el ciclo del agente: percibir -> decidir -> actuamos -> iterar

def agente_simple(objetivo, max_iteraciones=3):
  """
  Agente conceptual que simula el ciclo PERCIBIR-DECIDIR-ACTUAR-ITERAR.
  Sin herramientas reales, solo demostrando la lógica
  """

  contexto = f"Objetivo: {objetivo}\n"

  for i in range (max_iteraciones):
    print(f"Iteración {i+1}")

    # El "cerebro" (LLM) decide qué hacer basado en el contexto actual
    prompt_decision=f"""
    {contexto}

    Eres un agente que debe cumplir el objetivo paso a paso.
    ¿Qué acción tomarías AHORA? Responde en un sola línea
    """

    decision = client.models.generate_content(
        model="gemini-3-flash-preview",
        contents= prompt_decision
    ).text.strip()

    print(f"Decisión del agente: {decision}")

    # Simular una "observación" (en un agente real, aquí vendría una herramienta)
    observacion = f"Acción'{decision}' ejecutada exitosamente"
    print(f"observación: {observacion}")

    # Actualizar el contexto (memoria a corto plazo)
    contexto += f"\nacción{i+1}: {decision}\nObservación: {observacion}"

    # Condición de parada (Simulada)
    if i >= 1:
      print("\n Objetivo alcanzado. Agente finaliza")
      break

  return contexto

# Ejecución del agente
resultado = agente_simple("Planificar una ruta de entrega eficiente para 3 paquetes")
````

````python
# Prompt ReAct para un agente que puede utilizar herramientas

prompt_react = """

SYSTEM:
Eres un agente de análisis energético, Tienes acceso a herramientas.

HERRAMIENTAS DISPONIBLES:
1. Calculadora: realiza operaciones matemáticas. Input: expresión matemática.
2. BaseDatosEnergetica: consulta el histórico de consumo. Input: período de tiempo.

FORMATO DE RESPUESTA (OBLIGATORIO):
Thought: [Tu razonamiento paso a paso]
Action: [nombre de la herramienta]
Action Input: [input para la herramienta]
...(repetir Thought/Action/Observation si es necesario)

Thought: Ahora tengo la respuesta final.
Final Answer: [respuesta al usuario]

EJEMPLO:
Usuario. ¿El consumo actual de 180" es anómalo si el promedio es 120W?

Thought: Necesito calcular la diferencial porcentual.
Action: Calculadora
Action Input: ((180-120)/120)*100
Observation: 50
Thought: El incremento es del 50%, supera el umbral del 30%, Debo generar una alerta.
Final Answer: El consumo de 180W es anómalo. Representa un incremento del 50% sobre el promedio de 120W.

USER INPUT:
Consumo actual: 135W. Promedio histórico: 150W. ¿Es anómalo?
"""

respuesta_react = client.models.generate_content(
        model="gemini-3-flash-preview",
        contents= prompt_react
)

print("Respuesta del agente (patrón ReAct): ")
print(respuesta_react.text)
````

# **Herramientas(Tools) con function calling**

Construcción de un agente real que puede ejecutar funciones.

````python
# Definición de herramientas (funciones en Python)

import json
from datetime import datetime

def calcular_porcentaje_incremento(actual: float, promedio: float) -> float:
  """
  Calcula el porcentaje de incremento porcentual entre el valor actual y el promedio.
  """
  if promedio == 0:
    return float('inf')
  return((actual-promedio)/promedio)*100

def obtener_hora_actual()-> str:
  """
  Devuelve la hora actual en formato HH:MM.
  """
  return datetime.now().strftime("%H:%M")

def consultar_historico(dias: int=7)->dict:
  """
  Simula la consulta de datos históricos de consumo energético
  """
  # Datos simulados (en producción, esto vendría de una base de datos)
  return{
      "dias":dias,
      "consumo_promedio": 120.0,
      "consumo_maximo": 250.0,
      "consumo_minimo": 80.0,
      "datos":[110,125,130,118,122,115,120]
  }

# Mapa de herramientas para ejecución
herramientas_disponibles = {
    "calcular_porcentaje_incremento": calcular_porcentaje_incremento,
    "obtener_hora_actual": obtener_hora_actual,
    "consultar_historico": consultar_historico
}

# Imprimir valores
print("Herramientas definidas: ")
for nombre in herramientas_disponibles:
  print(f" -{nombre}")
````

````python
# Agente con herramientas en acción

from google.genai import types

def ejecutar_agente_con_herramientas(pregunta_usuario: str):
  """
  Ejecuta un agente que puede razonar y usar herramientas iterativamente.
  """

  # Historia de la conversación
  historial = [
      types.Content(role="user", parts=[types.Part.from_text(text=pregunta_usuario)])
  ]

  print(f"\n{'='*60}")
  print(f"Usuario: {pregunta_usuario}")
  print(f"{'='*60}")

  for iteracion in range(5): # máximo 5 iteraciones
    print(f"\nIteración {iteracion+1}")

    # Llamada a Gemini con herramientas disponibles
    respuesta_react = client.models.generate_content(
          model="gemini-3-flash-preview",
          contents= historial,
          config=types.GenerateContentConfig(
              tools=[herramientas_gemini],
              system_instruction = "Eres un agente supervisor energético. Usa herramientas cuando sea necesario para responder con precisión. No inventes datos."
          )
  )
    # Verificar si Gemini quiere utilizar una herramienta
    if respuesta.function_calls:
      for fc in respuesta.automatic_function_calls:
        nombre_func = fc.name
        args = fc.args

        print(f"El agente quiere usar la función: {nombre_func}")
        print(f"Argumentos: {args}")

        # Ejecutar la herramienta real
        if nombre_func in herramientas_disponibles:
          try:
            resultado = herramientas_disponibles[nombre_func](**args)
            print(f" Resultados:{resultado}")

            # Agregar el resultado al historial como observación
            historial.append(types.Content(
                role="tool",
                parts=[types.Part.from_function_call(name=nombre_func, args=args)]
                ))
          except Exception as e:
              print(f"Error al ejecutar la herramienta: {e}")
    else:
          # No hay más herramientas por utilizar, respuesta final
          print(f"\n Respuesta final del agente: ")
          print(respuesta.text)
          return respuesta.text
    return "El agente alcanzó el límite de iteraciones"

  # Probar el agente con una pregunta que requiere herramientas
  resultado= ejecutar_agente_con_herramientas(
      "El consumo actual es 180W. Consulta el histórico y dime si es anómalo comparado con el promedio."
  )
````

# **Memoria del agente**

Un agente sin memoria es una conversación aislada. La memoria permite recordar, mantener contexto, aprender y consultar conocimiento histórico (Módulo #2, diapositivas 34-35).

* ## Memoria de corto plazo (context window)
* ## Memoria de largo plazo (embeddings)


````python
"""
Memoria de corto plazo (context window)

El agente recuerda información dentro de la misma conversación

"""

def agente_con_memoria_corto_plazo():
  """
  Simula un chatbot con memoria de corto plazo.
  La memoria se mantiene en el historial de la conversación.
  """

  historial = []

  print("Agente con memoria de corto plazo")
  print("Escribe 'salir' para terminar\n")

  # Simular 3 turnos de conversación
  conversaciones = [
      "Mi nombre es David y soy profesor de la clase: Optativa III de Agentes inteligentes",
      "¿Cómo me llamo y de qué clase soy profesor?",
      "¿Qué te dije que estaba haciendo"
  ]

  for mensaje_usuario in conversaciones:
    print(f"Usuario: {mensaje_usuario}")
    # Agregar mensaje del usuario al historial
    historial.append(f"Usuario: {mensaje_usuario}")

    # Construir prompt con todo el historial (memoria a corto plazo)
    prompt = "\n".join(historial) + "\nAgente:"

    respuesta = client.models.generate_content(
        model="gemini-3-flash-preview",
        contents= f"Eeres un asistente con memoria. Responde basándote en la conversación previa.\n\n{historial}\nAgente:"
)
    print(f"Agente: {respuesta.text}")

    # Agregar respuesta del agente al historial
    historial.append(f"Agente: {respuesta.text}")

  print("Nota: Si la conversación excede el límite de tokens, el contexto antiguo se pierde.")
  print(" Esto es la LIMITACIÓN de la ventana de contexto (m2 d39)")
agente_con_memoria_corto_plazo()
````

````python
memoria_largo_plazo=[]

def almacenar_en_memoria(texto: str):
  """ Almacena un texto en la memoria de largo plazo con su embedding """
  resultado = client.models.embed_content(
      model="gemini-3-flash-preview",
      contents=texto,
      config= types.EmbedContentConfig(task_type="RETRIEVAL_DOCUMENT")
  )
  embedding=resultado.embedding[0].values

  memoria_largo_plazo.append({
      "texto": texto,
      "embedding": embedding
  })
  print(f"Texto almacenado: {texto[:50]}...")

def buscar_en_memoria(consulta: str, top_k:int =2):
  """
  Busca el texto más similar en la memoria de largo plazo
  """
  # Convertir consulta a embedding
  resultado = client.models.embed_content(
      model="gemini-3-flash-preview",
      contents=consulta,
      config= types.EmbedContentConfig(task_type="RETRIEVAL_QUERY")
  )
  query_embedding=np.array(resultado.embedding[0].values)

  # Calcular similitud coseno con cada elemento de la memoria
  similitudes = []
  for item in memoria_largo_plazo:
    emb = np.array(item["embedding"])
    similitud = np.dot(query_embedding, emb) / (np.linalg.norm(query_embedding) * np.linalg.norm(emb))
    similitudes.append(similitud,item["texto"])

  # Ordenar por similitud descendente
  similitudes.sort(reverse= Tre, key= lambda x: x[0])

  print(f"\n Búsqueda: '{consulta}'")
  for similitud, texto in similitudes[:top_k]:
    print(f"Similitud: {similitud:.4f} - Texto: {texto[:50]}...")

  return [texto for _, texto in similitud[:top_k]]

  # Almacenar conocimiento en la memoria de largo plazo
  print("Almacenado conocimiento en la memoria de largo plazo...\n")
  almacenar_en_memoria("El consumo actual es 180W. Consulta el histórico y dime si es anómalo comparado con el promedio.")
  almacenar_en_memoria("El umbral de alerta por consumo anómalo es un incremento superior al 30%")

  # Buscar en la memoria
  contexto_recuperado = buscar_en_memoria("¿Cuál es el umbral para detectar anomalías")
````

**Integración de memoria en el prompt**

````python
# Agente que usa memoria a largo plazo para responder

def agente_con_memoria_largo_plazo(pregunta_usuario: str):
  """
  Agente que recupera contexto de la memoria de largo plazo
  y lo inyecta en el prompt antes de responder.
  """
  # 1. Recuperar contexto relevante de la memoria
  contexto_recuperado= buscar_en_memoria(pregunta, top_k=2)
  contexto = "\n".join([f" -{c}" for c in contexto_recuperado])

  # 2. Construir prompt con contexto recuperado
  prompt_aumentado = f"""
  CONTEXTO RECUPERADO DE LA MEMORIA:
  {contexto}

  PREGUNTA DEL USUARIO:
  {pregunta}

  Responde basándote ÚNICAMENTE en el contexto proporcionado.
  Si el contexto no contiene la información necesaria, di "No tengo información suficiente".
  """

  # 3. Generar respuesta
  respuesta = client.models.generate_content(
      model="gemini-3-flash-preview",
      contents= prompt_aumentado
  )

  print(f"\n Pregunta: {pregunta}")
  print(f"Contexto recuperado:\n{contexto}")
  print(f"\n Respuesta.{respuesta.text}")

  return respuesta.text

# Probar el agente con memoria de largo plazo
agente_con_memoria_largo_plazo("¿Cuándo debo generar una alerta de consumo?")
````

````python
# System prompt + constraints, ahora con detección de patrones por hora del día

prompt_sistema_horario = """

SYSTEM:
Eres un agente supervisor energético experto en IoT.
Tu misión es analizar métricas de consumo, detectar anomalías y reconocer patrones de consumo típicos según la hora del día.

CONSTRAINTS:
- No inventes datos bajo ninguna circunstancia.
- Si faltan datos (incluida la hora), solicita información adicional.
- Responde exclusivamente en formato JSON válido.
- Usa solamente las herramientas proporcionadas.

REASONING:
Razonar paso a paso antes de decidir:
1. Compara el consumo actual contra el promedio histórico general.
2. Ubica la hora reportada dentro de una franja horaria (madrugada 00-06, mañana 06-12, tarde 12-18, noche 18-24).
3. Evalúa si el consumo es coherente con el patrón esperado para esa franja (por ejemplo, picos en horas de mañana/noche por uso doméstico, valles en madrugada).
4. Combina ambos análisis (histórico + horario) para decidir el estado final.

OUTPUT FORMAT:
{
  "status": "OK | ALERT",
  "analysis": "descripción del análisis comparando consumo actual vs promedio histórico",
  "hora_reportada": "HH:MM",
  "franja_horaria": "madrugada | mañana | tarde | noche",
  "patron_horario": "descripción de si el consumo es coherente o no con el patrón esperado para esa franja",
  "recommended_action": "acción sugerida",
  "confidence": 0.0 a 1.0
}

FEW-SHOT EXAMPLE:
Entrada: Consumo actual 200W, promedio 120W, hora 03:00
Salida: {"status":"ALERT", "analysis": "Incremento superior al 66% respecto al promedio histórico", "hora_reportada": "03:00", "franja_horaria": "madrugada", "patron_horario": "Anómalo: en madrugada se espera un consumo bajo por ausencia de actividad, pero se registra un pico", "recommended_action": "Reducir carga no crítica y verificar dispositivos activos de forma inusual en horario nocturno", "confidence": 0.95}

USER INPUT:
Consumo actual 180W, promedio histórico 120W, hora 08:30
"""

respuesta_horario = client.models.generate_content(
        model="gemini-3-flash-preview",
        contents=prompt_sistema_horario
)

print("Respuesta del agente con patrón horario: ")
print(respuesta_horario.text)
````

### ¿Cómo cambia el formato JSON de salida?

Respecto al formato original (`status`, `analysis`, `recommended_action`, `confidence`), se agregan **tres campos nuevos** derivados de incorporar la hora del día como variable de análisis:

| Campo nuevo | Propósito |
|---|---|
| `hora_reportada` | Deja explícita la hora sobre la que se está razonando (trazabilidad del dato de entrada). |
| `franja_horaria` | Clasifica la hora en una franja (madrugada/mañana/tarde/noche) para poder aplicar reglas de patrón esperado. |
| `patron_horario` | Comunica si el consumo es coherente o no con lo esperado para esa franja horaria, de forma independiente al análisis contra el promedio histórico. |

Además, `status` y `analysis` dejan de depender **únicamente** del promedio histórico: ahora `status` puede pasar a `ALERT` incluso si el consumo es "normal" en términos absolutos, si resulta incoherente con la franja horaria (ej. un pico de consumo a las 3 a.m.). Esto obliga al agente a razonar sobre dos señales combinadas en vez de una sola.

## Tarea 2 — Herramienta `enviar_alerta` con function calling

Se agrega una nueva herramienta que simula el envío de una notificación a un destinatario, y se integra al agente de function calling junto con las herramientas ya existentes (`calcular_porcentaje_incremento`, `obtener_hora_actual`, `consultar_historico`).

**Nota:** el bloque original `ejecutar_agente_con_herramientas` (celda `cqBVC-xH9j73`) tenía referencias a una variable `herramientas_gemini` nunca definida y a `respuesta` en vez de `respuesta_react`. Se corrigen esos errores aquí para que el agente realmente funcione.

````python
# Nueva herramienta: enviar_alerta

def enviar_alerta(destinatario: str, mensaje: str) -> dict:
  """
  Simula el envío de una notificación/alerta a un destinatario.

  Args:
    destinatario: nombre o identificador de quien recibe la alerta (ej. "operador_turno_noche").
    mensaje: contenido de la notificación a enviar.
  """
  timestamp = datetime.now().strftime("%Y-%m-%d %H:%M:%S")
  print(f"[ALERTA ENVIADA] -> {destinatario}: {mensaje} ({timestamp})")
  return {
      "destinatario": destinatario,
      "mensaje": mensaje,
      "estado": "enviado",
      "timestamp": timestamp
  }

# Se registra la nueva herramienta junto con las existentes
herramientas_disponibles["enviar_alerta"] = enviar_alerta

# Lista de funciones que se exponen a Gemini para function calling
herramientas_gemini = [
    calcular_porcentaje_incremento,
    obtener_hora_actual,
    consultar_historico,
    enviar_alerta,
]

print("Herramientas disponibles para el agente: ")
for nombre in herramientas_disponibles:
  print(f" -{nombre}")
````

````python
# Agente con herramientas en acción (versión corregida, incluye enviar_alerta)

def ejecutar_agente_con_herramientas(pregunta_usuario: str):
  """
  Ejecuta un agente que puede razonar y usar herramientas iterativamente,
  incluyendo el envío de alertas cuando detecta una anomalía.
  """

  historial = [
      types.Content(role="user", parts=[types.Part.from_text(text=pregunta_usuario)])
  ]

  print(f"\n{'='*60}")
  print(f"Usuario: {pregunta_usuario}")
  print(f"{'='*60}")

  for iteracion in range(5):  # máximo 5 iteraciones
    print(f"\nIteración {iteracion+1}")

    respuesta_react = client.models.generate_content(
        model="gemini-3-flash-preview",
        contents=historial,
        config=types.GenerateContentConfig(
            tools=herramientas_gemini,
            system_instruction=(
                "Eres un agente supervisor energético. Usa las herramientas disponibles "
                "(incluida enviar_alerta cuando detectes una anomalía) para responder con "
                "precisión. No inventes datos."
            ),
        ),
    )

    if respuesta_react.function_calls:
      for fc in respuesta_react.function_calls:
        nombre_func = fc.name
        args = fc.args

        print(f"El agente quiere usar la función: {nombre_func}")
        print(f"Argumentos: {args}")

        if nombre_func in herramientas_disponibles:
          try:
            resultado = herramientas_disponibles[nombre_func](**args)
            print(f" Resultado: {resultado}")

            # Se agrega la llamada a función y su resultado al historial
            historial.append(types.Content(
                role="model",
                parts=[types.Part.from_function_call(name=nombre_func, args=args)]
            ))
            historial.append(types.Content(
                role="user",
                parts=[types.Part.from_function_response(
                    name=nombre_func, response={"resultado": resultado}
                )]
            ))
          except Exception as e:
            print(f"Error al ejecutar la herramienta: {e}")
    else:
      # No hay más herramientas por utilizar, respuesta final
      print(f"\n Respuesta final del agente: ")
      print(respuesta_react.text)
      return respuesta_react.text

  return "El agente alcanzó el límite de iteraciones"


# Probar el agente con una pregunta que requiere consultar histórico y enviar una alerta
resultado = ejecutar_agente_con_herramientas(
    "El consumo actual es 180W. Consulta el histórico y dime si es anómalo comparado con el "
    "promedio. Si es anómalo, envía una alerta al destinatario 'operador_turno_noche' "
    "explicando brevemente la anomalía."
)
````

## Tarea 4 — Depuración de un prompt ReAct defectuoso

Prompt a analizar:

```python
prompt_defectuoso = """
Eres un agente que responde preguntas. Puedes usar herramientas.
Usuario: ¿Cuánto es 25*4+160?
Piensa y responde
"""
```

### Problemas encontrados

1. **Falta de herramientas definidas**: dice "puedes usar herramientas" pero no declara ninguna (nombre, qué hace, qué input espera). El modelo no tiene forma de saber qué puede invocar ni cómo.
2. **Falta de formato de salida obligatorio**: no exige la estructura `Thought / Action / Action Input / Observation / Final Answer`. Sin ese formato, la respuesta no se puede parsear con código para ejecutar herramientas reales; el modelo simplemente conversa en texto libre.
3. **Falta de ejemplos (few-shot)**: no muestra ningún caso resuelto que enseñe al modelo cómo alternar razonamiento y acción, por lo que no tiene una plantilla clara a seguir.
4. **Mezcla de roles sin delimitar**: no separa claramente instrucción de sistema (SYSTEM) y entrada del usuario (USER INPUT); todo está en un bloque de texto plano, lo que puede confundir qué parte es instrucción y cuál es la pregunta a responder.
5. **Instrucción final ambigua**: "Piensa y responde" no aclara si es obligatorio usar una herramienta (p. ej. una calculadora) en vez de calcular mentalmente, ni cuándo debe detenerse el razonamiento.

````python
# Versión original (defectuosa) vs versión corregida del prompt ReAct

prompt_defectuoso = """
Eres un agente que responde preguntas. Puedes usar herramientas.
Usuario: ¿Cuánto es 25*4+160?
Piensa y responde
"""

prompt_corregido = """

SYSTEM:
Eres un agente de razonamiento que responde preguntas. Tienes acceso a herramientas externas y
debes usarlas para cualquier cálculo, en lugar de calcular mentalmente.

HERRAMIENTAS DISPONIBLES:
1. Calculadora: realiza operaciones matemáticas. Input: expresión matemática (ej: "25*4+160").

FORMATO DE RESPUESTA (OBLIGATORIO):
Thought: [tu razonamiento paso a paso]
Action: [nombre de la herramienta]
Action Input: [input para la herramienta]
Observation: [resultado de la herramienta]
...(repetir Thought/Action/Action Input/Observation cuantas veces sea necesario)
Thought: Ahora tengo la respuesta final.
Final Answer: [respuesta final al usuario]

EJEMPLO:
Usuario: ¿Cuánto es 10*2+5?
Thought: Necesito calcular esta expresión aritmética usando la calculadora.
Action: Calculadora
Action Input: 10*2+5
Observation: 25
Thought: Ya tengo el resultado final.
Final Answer: El resultado de 10*2+5 es 25.

USER INPUT:
¿Cuánto es 25*4+160?
"""

respuesta_defectuosa = client.models.generate_content(
    model="gemini-3-flash-preview",
    contents=prompt_defectuoso
)

respuesta_corregida = client.models.generate_content(
    model="gemini-3-flash-preview",
    contents=prompt_corregido
)

print("=== Respuesta con prompt DEFECTUOSO ===")
print(respuesta_defectuosa.text)

print("\n=== Respuesta con prompt CORREGIDO ===")
print(respuesta_corregida.text)
````

### Comparación de resultados

Con la aritmética simple del ejemplo (25*4+160=260), ambos prompts probablemente lleguen al **mismo resultado numérico**, porque el modelo puede resolverlo mentalmente sin necesitar la herramienta. La diferencia real está en la **forma** de la respuesta y en qué tan segura/reutilizable es esa respuesta dentro de un sistema automatizado:

| Aspecto | Prompt defectuoso | Prompt corregido |
|---|---|---|
| Formato de salida | Texto libre, sin estructura fija | Estructura fija `Thought/Action/Action Input/Observation/Final Answer` |
| Uso explícito de herramienta | No se sabe si "razonó" o "calculó mentalmente" | Declara la `Action: Calculadora` y su input antes de responder |
| Parseable por código (para ejecutar herramientas reales) | No — no hay forma de extraer una acción/input con un parser | Sí — se puede extraer `Action` y `Action Input` con una expresión regular simple |
| Riesgo de alucinación en cálculos complejos | Alto (nada obliga a delegar el cálculo a una herramienta real) | Bajo (el formato empuja a delegar el cálculo a la Calculadora en cada paso) |
| Reutilizable en un pipeline con ejecución real de herramientas | No | Sí |
| Presencia de ejemplo (few-shot) que guíe el formato | No | Sí |

**Conclusión:** el prompt defectuoso puede "acertar" en preguntas triviales porque el modelo resuelve el cálculo internamente, pero no sirve como base de un agente ReAct real: no hay contrato de formato, no hay herramientas declaradas y no hay ejemplo que fije el patrón. El prompt corregido sí es apto para integrarse con un bucle de ejecución de herramientas como el usado en `ejecutar_agente_con_herramientas`.
